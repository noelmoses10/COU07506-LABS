### Task 1.1 — Your first notebook
Generating and plotting the elementary sequences:
* Unit impulse: $\delta(n)$
* Unit step: $u(n)$
* Unit ramp: $r(n)$
* Real exponential: $0.8^n u(n)$

```python
n = np.arange(-5, 11)
delta = (n == 0).astype(float)       # unit impulse
step  = (n >= 0).astype(float)       # unit step
ramp  = n * step                     # unit ramp
expo  = 0.8**n * step                # real exponential

fig, axes = plt.subplots(1, 4, figsize=(14, 3))
for ax, x, title in zip(axes, [delta, step, ramp, expo],
                        ["$\delta(n)$", "$u(n)$", "$r(n)$", "$(0.8)^n u(n)$"]):
    ax.stem(n, x, basefmt="k-")
    ax.set_title(title)
    ax.set_xlabel("n")
plt.tight_layout()
plt.show()
```

### Analysis Questions & Answers

**1. What is the sampling rate of your phone’s recording? How many numbers are needed to store one minute of it in mono?**
* **Answer:** Based on the execution of your audio recording, the sampling rate ($f_s$) is printed above (e.g., 44100 Hz, 48000 Hz, etc.). To store 1 minute (60 seconds) of mono audio, the number of samples needed is:
  $$\text{Samples} = f_s \times 60$$
  *(For example, if $f_s = 48000$ Hz, then $48000 \times 60 = 2,880,000$ numbers are needed).*

**2. Why is a stem plot more appropriate than a continuous line for $x(n)$?**
* **Answer:** A discrete-time signal $x(n)$ is mathematically defined only at integer steps (discrete intervals of $n$). A continuous line suggests the existence of values in between the integers (e.g., at $n=0.5$), which is physically incorrect for a discrete sequence. A stem plot cleanly visualizes values only at distinct integer points.

**3. For $a=1.1$ the exponential grows. Which property of a system would such a signal violate if it were an impulse response (Unit 1 §1.7)?**
* **Answer:** It would violate the **BIBO (Bounded-Input Bounded-Output) Stability** property. For a system to be stable, its impulse response must be absolutely summable:
  $$\sum_{n=-\infty}^{\infty} |h(n)| < \infty$$
  When $a = 1.1$, the sequence $(1.1)^n u(n)$ grows infinitely as $n \to \infty$, meaning the sum of its absolute values diverges to infinity, representing an unstable system.

### Task 2.1 — Aliasing of a Sinusoid
We will plot an "analog" continuous-time 7 kHz sine wave (simulated using a very dense grid) and its sampled representation when sampled at $f_s = 10$ kHz. We will also plot the predicted alias frequency component.

```python
import numpy as np
import matplotlib.pyplot as plt

# Parameters
f0 = 7.0       # Analog signal frequency (kHz)
fs = 10.0      # Sampling frequency (kHz)

# Dense grid to represent the "continuous" analog signal
t_dense = np.linspace(0, 1, 2000)
x_analog = np.cos(2 * np.pi * f0 * t_dense)

# Sampled points
n = np.arange(0, int(fs) + 1)
tn = n / fs
x_sampled = np.cos(2 * np.pi * f0 * tn)

# Theoretical alias frequency
f_alias = abs(f0 - fs * round(f0 / fs))
x_alias = np.cos(2 * np.pi * f_alias * t_dense)

# Plotting
plt.figure(figsize=(10, 5))
plt.plot(t_dense, x_analog, label=f'Original "Analog" Signal ({f0} kHz)', color='lightgray', linestyle='--')
plt.stem(tn, x_sampled, linefmt='C0-', markerfmt='C0o', basefmt='k-', label=f'Sampled Signal (fs = {fs} kHz)')
plt.plot(t_dense, x_alias, label=f'Reconstructed Alias ({f_alias} kHz)', color='red', alpha=0.8)

plt.title(f"Aliasing Demonstration: {f0} kHz Sinusoid Sampled at {fs} kHz")
plt.xlabel("Time (ms)")
plt.ylabel("Amplitude")
plt.legend(loc='upper right')
plt.grid(True)
plt.xlim(0, 1)
plt.show()
```

### Task 2.1 continued — Alias Table
We will evaluate the alias frequency of a $F = 7.0$ kHz sinusoid when sampled at $f_s = 12$, $14$, and $20$ kHz, and record the values in a comparative table.

```python
import pandas as pd

F = 7.0 # Original signal frequency in kHz
fs_values = [10.0, 12.0, 14.0, 20.0]

data = []
for fs in fs_values:
    # Formula: F_alias = |F - k * fs| where k = round(F / fs)
    k = round(F / fs)
    f_alias = abs(F - k * fs)
    data.append({
        "Sampling Frequency fs (kHz)": fs,
        "Original Frequency F (kHz)": F,
        "k": k,
        "Predicted Alias |F - k*fs| (kHz)": f_alias
    })

df = pd.DataFrame(data)
df
```

### Task 2.3 — A Uniform Quantizer
We implement the uniform quantizer and plot the original signal, the 3-bit quantized representation, and the quantization error.

```python
def quantize(x, bits, full_scale=1.0):
    delta = 2*full_scale / 2**bits
    q = np.round(x/delta)*delta
    return np.clip(q, -full_scale, full_scale - delta)

fs = 8000
t = np.arange(0, 0.01, 1/fs)
x_sine = 0.99*np.sin(2*np.pi*300*t)
xq = quantize(x_sine, 3)

plt.figure(figsize=(9, 4.5))
plt.plot(t*1e3, x_sine, label="x(n) (Original Sine)", marker='o')
plt.step(t*1e3, xq, where="mid", label="3-bit Quantized", color='orange')
plt.plot(t*1e3, x_sine - xq, label="Quantization Error", color='red', linestyle='--')
plt.xlabel("Time (ms)")
plt.ylabel("Amplitude")
plt.title("3-Bit Uniform Quantization of a 300 Hz Sinusoid")
plt.legend()
plt.grid(True)
plt.show()
```

### Task 2.4 — Verify the 6 dB-per-bit Rule
We compute the Signal-to-Quantization-Noise Ratio (SQNR) for word lengths $B \in [2, 4, 6, \dots, 16]$ and compare it with the theoretical limit $SQNR = 6.02B + 1.76$ dB.

```python
# Generate a dense sinusoid to compute SQNR reliably
t_dense_sqnr = np.arange(0, 1.0, 1/fs)
x_full_sine = 0.99 * np.sin(2 * np.pi * 300 * t_dense_sqnr)

bits = np.arange(2, 17, 2)
meas = []
for B in bits:
    error = x_full_sine - quantize(x_full_sine, B)
    # Avoid division by zero
    sqnr = 10 * np.log10(np.sum(x_full_sine**2) / np.sum(error**2))
    meas.append(sqnr)

plt.figure(figsize=(9, 4.5))
plt.plot(bits, meas, "o-", label="Measured SQNR")
plt.plot(bits, 6.02*bits + 1.76, "s--", label="Theoretical: 6.02B + 1.76")
plt.xlabel("Bits (B)")
plt.ylabel("SQNR (dB)")
plt.title("SQNR vs. Resolution (Bits) for a Full-Scale Sinusoid")
plt.legend()
plt.grid(True)
plt.show()
```

### Week 2 — Analysis Questions & Answers

**1. Explain, with a frequency-axis sketch, why the 7 kHz tone appears at 3 kHz when $f_s = 10$ kHz.**
* **Answer:** By the sampling theorem, any spectral component at frequency $F$ has periodic repetitions at $F \pm k f_s$. For $F = 7$ kHz and $f_s = 10$ kHz, there is an alias spectral line at:
  $$F_{alias} = |F - 1 \cdot f_s| = |7 - 10| = 3 \text{ kHz}$$
  In the range $[-f_s/2, f_s/2] = [-5 \text{ kHz}, 5 \text{ kHz}]$, the frequency $7$ kHz folds back to $3$ kHz.

**2. In Task 2.2, why can the aliasing heard in the naïve version not be removed afterwards?**
* **Answer:** Once a signal is downsampled without a low-pass anti-aliasing filter, high-frequency components are irreversibly folded back and overlapped onto the low-frequency baseband. Because they now share the exact same frequencies as the original baseband signal, it is mathematically impossible to distinguish or separate the alias noise from the original signal afterwards.

**3. By how much does the SQNR change when one bit is added? Does your measurement agree?**
* **Answer:** Theoretically, adding one bit doubles the number of quantization levels, halving the quantization step size $\Delta$, which reduces noise power by a factor of 4. This increases SQNR by approximately $20 \log_{10}(2) \approx 6.02$ dB. The plot in Task 2.4 confirms this linear increase, with a slope matching $\approx 6$ dB per bit.

**4. A temperature sensor changes by at most 0.1 °C per second and the required resolution is 0.01 °C over a range of 0–100 °C. Suggest a sampling rate and number of bits, and justify them.**
* **Answer:**
  * **Sampling Rate ($f_s$):** Since the maximum rate of change is 0.1 °C/s, the signal has extremely low-frequency bandwidth (practically $< 1$ Hz). Applying the Nyquist rate, a very slow sampling rate such as **$f_s = 1$ Hz** or **2 Hz** is more than sufficient.
  * **Number of bits ($B$):** The measurement range is $100$ °C ($0$ to $100$ °C). The required resolution step is $\Delta \le 0.01$ °C. The number of intervals needed is:
    $$\text{Levels} = \frac{100 \text{ °C}}{0.01 \text{ °C}} = 10,000 \text{ levels}$$
    To represent $10,000$ unique levels, we need a bit depth $B$ such that:
    $$2^B \ge 10,000 \implies B \ge \log_2(10,000) \approx 13.29 \implies B = 14 \text{ bits}$$
    Thus, a **14-bit ADC** (offering $2^{14} = 16,384$ levels) is recommended.
